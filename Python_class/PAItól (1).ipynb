**Task** : Selecting DataFrame columns

```python
import pandas as pd
Students = pd.DataFrame({
    "Name" : ["Anna", "David", "Emma"],
    "Age" : [20,25,22],
    "City" : ["London", "Paris", "Berlin"]})

#Print the entire student name
print(Students["Name"])

#Print the entire student city
print(Students["City"])

#Print the student name : Anna
print(Students["Name"][0])
```
