```python

import pandas as pd
#1. Daten abholen (Extract)
def extract_data() :
  #Wir simulieren hier eine 'schmutzige' Rohfatei
  raw_data = {
      'Product' : ['Laptop', 'Maus', 'Tastatur', 'Monitor'],
      'Preis_String' : ['800E', '40Euro', '200'], #Chaos! Text statt Zahlen
      'Verkauft' : [4,5,7,6]
  }
  df = pd.DataFrame(raw_data)
  print('1. Daten erfolgreich extrahiert (geholt).')
  return df

#2. Transform (Daten putzen)
def transform_data(df):
  # Problem : 'Preis_String' ist Text (wegen dem E Zeichen). Ich will damit nich rechnen.
  # a). Wir entfernen das 'E' und 'Euro'
  # Wir erstzen 'E' durch nichts ''
  df['Preis_String'] = df['Preis_String'].str.replace('E', '').str.replace('Euro', '').str.replace(' ', '')

  # b). wir wandeln den Text in Zahlen (Float) um
  df['Preis_Zahl'] = df['Preis_String'].astype(float)

  # c). Wir berechnen den Umsatz (Preis * Menge)
  df['Umsatz'] = df['Preis_Zahl'] * df['Verkauft']

  # d). Wir löschen die alte schmutzige Spalte
  df_sauber = df.drop(columns=['Preis_String'])

  print('2. Date erfolgreich transformiert (geputzt).')
  return df_sauber

  #3. Load(Daten speichern)
  def load_data(df):
    #Wir speichern die saubere Datei als CSV

    df.to_csv('saubere_umsatzdaten.csv', index=False)
    print('3. Daten erfolgreich geladen(gespeichert).')

  #Alles verbinden
  def run_etl():
    print('__Start ETL prozess__')

    data = extract_data() #holen
    clean_data = transform_data(data) #putzen
    load_data(clean_data) #speichern

    print('__ ETL Prozess fertig!__')
    return clean_data

  #Wir starten die Maschine

  ergebnis = run_etl()
  display(ergebnis)
```
