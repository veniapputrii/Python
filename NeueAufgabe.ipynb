```python
#12. März 2026
import pandas as pd
import matplotlib.pyplot as plt
#Unsere neuen Mitarbeiter-Daten inklusive Alter!

daten= {
    'Name' : ['Lukas', 'Mia', 'Noah', 'Emma', 'Paul', 'Lea', 'Finn', 'Marie', 'Leon', 'Sofia'],
    'Abteilung' : ['IT', 'HR', 'IT', 'HR', 'IT', 'HR', 'IT', 'HR', 'IT', 'HR'],
    'Alter' : [28, 34, 29, 42, 50, 25, 38, 31, 24, 45],
    'Gehalt' : [60000, 55000, 65000, 58000, 75000, 62000, 70000, 50000, 68000, 52000]
}
df_team = pd.DataFrame(daten)
print(df_team)

#-- Schritt 1 : Filtern (Nur Mitarbeiter strikt über 30 Jahre) --
#wir erstellen die Maske und wenden sie an
df_erfahren = df_team[df_team['Alter'] > 30]
print("1. Erfahrene Mitarbeiter (Ü30):")
print('_' * 30)

#-- Schritt 2 : Gruppieren & Rechnen (Gesamtsumme der Gehälter pro ABteilung)
#Wir gruppieren nach Abteilung und berechnen die Summe (.sum()) vom Gehalt
kosten_pro_Abteilung = df_erfahren.groupby('Abteilung')['Gehalt'].sum()
print('2. Gehaltskosten der erfahrenen Mitarbeiter pro Abteilung :')
print(kosten_pro_Abteilung)
print('_' * 30)
```
