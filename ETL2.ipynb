```python
#2. Märch 2026
import pandas as pd
import matplotlib.pyplot as plt

#Teil 1 : eine Datei in Colab simulieren und speichern
test_daten = {
    'Name' : ['Anna', 'Ben', 'Clara', 'David', 'Emil', 'Fiona', 'Gutav', ''],
    'Abeilung' : ['IT', 'Sales', 'IT', 'Marketing', 'Sales', 'IT', 'Marketin','HR'],
    'Gehalt' : [55000, 48000, 62000, 45000, 52000, 59000, 41000, 39000]
}

#Speichert die Datei i der virtuellen Umbegung von Colab
pd.DataFrame(test_daten).to_csv('mitarbeiter_daten.csv', index = False)

#Datei direkt wieder einlasen
df_geladen = pd.read_csv('mitarbeiter_daten.csv')
print('1. Daten erfolgreich aus der CSV-Datei geladen!/n')

#Teil 2 : Daten sortieren
df_sortiert = df_geladen.sort_values(by='Gehalt', ascending=False)
print('2. Top 3 Verdiener:')
print(df_sortiert.head(3))
print('/n' + '-'*30 + '/n')

gehalt_pro_abteilung = df_geladen.groupby('Abeilung')['Gehalt'].sum()

#Teil 3 : das Diagram
gehalt_pro_abteilung.plot(kind='bar', color='skyblue')
plt.xlabel('Abteilung')
plt.ylabel('Gesamtgehalt')
plt.title('Gesamtgehalt pro Abteilung')
plt.ylabel('Gehalt in Euro')
plt.xlabel('Abtelung')
plt.tight_layout()
plt.show()
```
