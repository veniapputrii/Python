```python
#15. Feb
import pandas as pd
#Tabelle 1. Die Mitarbeiter
daten_m = {
    'Name' : ['Anna', 'Ben', 'Clara', 'David'],
    'Team_Code' : [1,2,1,3] #Das ist der Schlüssel
}

df_mitarbeitr = pd.DataFrame(daten_m)

#Tabelle 2 : Die Team-Namen
daten_t = {
    'ID' : [1,2,3],
    'Team_Name' : ['IT', 'Sales', 'HR'] #Die Übersetzung für den Code
}

df_teams = pd.DataFrame(daten_t)

print("___Tabelle 1: Mitarbeiter ___")
display(df_mitarbeitr)
print("___Tabelle 2: Teams ___")
display(df_teams)
```

```python
# 16.Feb
df_komplett = pd.merge(
    df_mitarbeitr,
    df_teams,
    left_on = 'Team_Code',
    right_on= 'ID'
)

display(df_komplett)
```

```python
#17. Feb
daten_m = {
    'Name' : ['Anna','Ben','Clara','David', 'Emil'],
    'Team_Code' : [1,2,1,3,99],
    'Stadt' : ['Berlin', 'Hamburg', 'München', 'Berlin', 'Frankfurt'],
    'ID'  : [1,2,3,4,5]
}
df_mitarbeiter_neu = pd.DataFrame(daten_m)
print(df_mitarbeiter_neu)
```

```python
#19. Feb. 2026
#Ich will ganz bestimmte Daten daraus filtern

df_komplett2 = pd.merge(
    df_komplett,
    df_mitarbeiter_neu,
    left_on = 'Team_Code',
    right_on = 'ID'
)

display(df_komplett2)
```

```python
#20. Feb. 2026
#Ich zeige euch alle Kunden aus Berlin
filter_Stadt = df_komplett2['Stadt'] == 'Berlin'

Berliner = df_komplett2[filter_Stadt]
display(Berliner)
```

```python
#UNd zeige euch alle Mitarbeiter aus Team '1'
filter_Team = df_komplett2['Team_Code_x'] == 1

Ersteteam = df_komplett2[filter_Team]
display(Ersteteam)
```
